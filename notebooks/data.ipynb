```python
import re

def clean_text(text):
    text = str(text)

    # Remove excessive whitespace
    text = re.sub(r'\s+', ' ', text)

    text = re.sub(r'[إأآ]', 'ا', text)
    text = text.replace('ى', 'ي')
    

    return text.strip()
```

```python
df["clean_text"] = df["text"].apply(clean_text)
```

```python


# --------------------------------------------------
# 1. Configuration
# --------------------------------------------------

CHUNK_SIZE = 1000
CHUNK_OVERLAP = 150
MIN_CHUNK_SIZE = 100



# --------------------------------------------------
# 3. Split long text into overlapping chunks
# --------------------------------------------------

def split_long_text(text, chunk_size=CHUNK_SIZE,
                    overlap=CHUNK_OVERLAP):

    if len(text) <= chunk_size:
        return [text] if text.strip() else []

    chunks = []
    start = 0

    while start < len(text):
        end = min(start + chunk_size, len(text))

        # Prefer ending at a sentence or line boundary
        if end < len(text):
            min_end = start + int(chunk_size * 0.7)

            candidates = [
                text.rfind("\n", min_end, end),
                text.rfind("۔", min_end, end),
                text.rfind(".", min_end, end),
                text.rfind("؛", min_end, end),
                text.rfind("،", min_end, end),
            ]

            valid_candidates = [pos for pos in candidates if pos >= min_end]

            if valid_candidates:
                end = max(valid_candidates) + 1

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        if end >= len(text):
            break

        # Move forward, keeping some overlap
        next_start = end - overlap

        # Prevent an infinite loop
        start = max(next_start, start + 1)

    return chunks


# --------------------------------------------------
# 4. Detect legal article boundaries
# --------------------------------------------------

ARTICLE_PATTERN = re.compile(
    r"(?m)^[ \t]*"
    r"(?:المادة|مادة)"
    r"[ \t]*(?:رقم[ \t]*)?"
    r"[\(\[（]?[ \t]*"
    r"[0-9٠-٩۰-۹]+"
    r"[ \t]*[\)\]）]?"
    r"[ \t]*(?:[-:ـ.]?[ \t]*)"
)


def split_into_articles(text):
    matches = list(ARTICLE_PATTERN.finditer(text)) 

    # If article boundaries cannot be identified reliably,
    # use generic chunking instead.
    if len(matches) < 2:
        return None

    sections = []

    # Keep any introduction before the first article
    if text[:matches[0].start()].strip():
        sections.append(text[:matches[0].start()].strip())

    for i, match in enumerate(matches):
        end = (
            matches[i + 1].start()
            if i + 1 < len(matches)
            else len(text)
        )

        article = text[match.start():end].strip()

        if article:
            sections.append(article)

    return sections


# --------------------------------------------------
# 5. Create chunks for each document
# --------------------------------------------------

def chunk_document(row, document_id):
    text = clean_text(row["text"])

    if not text:
        return []

    law_name = str(row.get("law_name", "Unknown law"))
    category = str(row.get("categories", "Unknown"))

    articles = split_into_articles(text)

    # Fallback when article structure is not detected
    sections = articles if articles else [text]

    chunks = []
    buffer = ""

    for section in sections:
        # Long article: split it into smaller chunks
        if len(section) > CHUNK_SIZE:
            if buffer.strip():
                chunks.append(buffer.strip())
                buffer = ""

            chunks.extend(
                split_long_text(section)
            )

        # Keep complete articles together when possible
        elif not buffer:
            buffer = section

        elif len(buffer) + len(section) + 2 <= CHUNK_SIZE:
            buffer += "\n\n" + section

        else:
            chunks.append(buffer.strip())
            buffer = section

    if buffer.strip():
        chunks.append(buffer.strip())

    results = []

    for chunk_index, chunk in enumerate(chunks):
        if len(chunk) < MIN_CHUNK_SIZE:
            # Merge tiny chunks with the preceding chunk
            if results:
                results[-1]["text"] += "\n\n" + chunk
                continue

        results.append({
            "id": f"doc_{document_id}_chunk_{chunk_index}",
            "text": chunk,
            "law_name": law_name,
            "category": category,
            "document_id": str(document_id),
            "chunk_index": chunk_index,
            "char_count": len(chunk),
        })

    return results


# --------------------------------------------------
# 6. Apply chunking to the dataset
# --------------------------------------------------

def create_chunks(df):
    all_chunks = []

    for document_id, (_, row) in enumerate(df.iterrows()):
        all_chunks.extend(
            chunk_document(row, document_id)
        )

    return pd.DataFrame(all_chunks)


```

```python
chunks_df = create_chunks(df)

chunks_df.to_csv(
        "../data/processed/legal_chunks.csv",
        index=False,
        encoding="utf-8-sig"
    )

print("Original documents:", len(df))
print("Generated chunks:", len(chunks_df))
print(chunks_df["char_count"].describe())

chunks_df.head(3)
```
